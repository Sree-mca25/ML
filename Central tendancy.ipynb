```python
import pandas as pd
import random

# Random number of records between 400 and 600
n = random.randint(400, 600)

# Create dataset
data = pd.DataFrame({
    "Screen_Time_Hours": [random.randint(1, 12) for i in range(n)],
    "Daily_Unlocks": [random.randint(5, 150) for i in range(n)],
    "Social_Media_Minutes": [random.randint(10, 400) for i in range(n)],
    "Sleep_Hours": [random.randint(4, 10) for i in range(n)],
    "Data_Usage_GB": [random.randint(1, 30) for i in range(n)]
})

# Save dataset
data.to_csv("smartphone_usage.csv", index=False)

# Central Tendency
x = data["Screen_Time_Hours"]

print("Total Records:", len(data))
print("Mean:", x.mean())
print("Median:", x.median())
print("Mode:", x.mode()[0])

# Dispersion
print("Standard Deviation:", x.std())
print("Variance:", x.var())
```
