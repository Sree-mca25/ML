```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 1. Load dataset
data = pd.read_csv("smartphone_usage.csv")

# 2. Input and output
# Target = Screen Time
y = data["Screen_Time_Hours"]


# ==========================================
# SIMPLE LINEAR REGRESSION
# ==========================================

# Take only Daily Unlocks
X = data[["Daily_Unlocks"]]

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=10
)

# Create and train model
model = LinearRegression()
model.fit(X_train, y_train)

# Predict
prediction = model.predict(X_test)

# Results
print("SIMPLE LINEAR REGRESSION")
print("MSE:", mean_squared_error(y_test, prediction))
print("R2 Score:", r2_score(y_test, prediction))

# Visualization
plt.scatter(X_test, y_test)
plt.plot(X_test, prediction)
plt.xlabel("Daily Unlocks")
plt.ylabel("Screen Time (Hours)")
plt.title("Simple Linear Regression")
plt.show()


# ==========================================
# MULTIPLE LINEAR REGRESSION
# ==========================================

# Take all input columns
X = data[
    [
        "Daily_Unlocks",
        "Social_Media_Minutes",
        "Data_Usage_GB",
        "Sleep_Hours"
    ]
]

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=10
)

# Create and train model
model = LinearRegression()
model.fit(X_train, y_train)

# Predict
prediction = model.predict(X_test)

# Results
print("\nMULTIPLE LINEAR REGRESSION")
print("MSE:", mean_squared_error(y_test, prediction))
print("R2 Score:", r2_score(y_test, prediction))

# Visualization
plt.scatter(y_test, prediction)
plt.xlabel("Actual Screen Time")
plt.ylabel("Predicted Screen Time")
plt.title("Multiple Linear Regression")
plt.show()
```
